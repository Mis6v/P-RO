##  Description du projet

Ce projet a pour objectif de modéliser et résoudre trois problèmes classiques et appliqués de Recherche Opérationnelle à l’aide de méthodes mathématiques et d’outils informatiques.

Les problèmes étudiés sont :

- La méthode du simplexe (programmation linéaire)
- Le problème des véhicules (VRP)
- Le transport de poissons entre Nouadhibou et Nouakchott
