# 1. Méthode du simplexe

## Description

Cette partie présente la résolution d’un problème de programmation linéaire avec la méthode du simplexe.

Le problème consiste à maximiser une fonction objectif sous plusieurs contraintes linéaires.

## Fonction objectif

Maximiser :


Z = 3X + 4Y


Sous les contraintes :


\begin{array}{l}
X + 2Y \leq 8 \\
3X + Y \leq 9 \\
X, Y \geq 0
\end{array}



## Méthode utilisée

Le problème est modélisé avec la bibliothèque PuLP.

---

```python
model = LpProblem("Simplexe", LpMaximize)

X = LpVariable("X", lowBound=0)
Y = LpVariable("Y", lowBound=0)

model += 3 * X + 4 * Y

model += X + 2 * Y <= 8
model += 3 * X + Y <= 9

model.solve()



print("Statut :", LpStatus[model.status])

print("X =", value(X))
print("Y =", value(Y))

print("Zmax =", value(model.objective))
```

# 2. Problème des véhicules (VRP)

## Description

Le problème des véhicules consiste à organiser les tournées d’un véhicule afin de minimiser la distance totale parcourue.

Dans cet exemple, un dépôt doit livrer plusieurs clients.

## Méthode utilisée

Le problème est résolu avec la bibliothèque OR-Tools développée par Google.

---

```python

distance_matrix = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 17],
    [15, 35, 0, 30, 28],
    [20, 25, 30, 0, 16],
    [25, 17, 28, 16, 0]
]

manager = pywrapcp.RoutingIndexManager(
    len(distance_matrix),
    1,
    0
)

routing = pywrapcp.RoutingModel(manager)

def distance_callback(from_index, to_index):

    from_node = manager.IndexToNode(from_index)
    to_node = manager.IndexToNode(to_index)

    return distance_matrix[from_node][to_node]

transit_callback_index = routing.RegisterTransitCallback(
    distance_callback
)

routing.SetArcCostEvaluatorOfAllVehicles(
    transit_callback_index
)

search_parameters = pywrapcp.DefaultRoutingSearchParameters()

search_parameters.first_solution_strategy = (
    routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
)

solution = routing.SolveWithParameters(search_parameters)

if solution:


    index = routing.Start(0)

    route = ""

    while not routing.IsEnd(index):

        route += f"{manager.IndexToNode(index)} -> "

        index = solution.Value(routing.NextVar(index))

    route += "0"

    print("Route optimale :")
    print(route)
```

# 3. Transport de poissons Nouadhibou -- Nouakchott

## Description

Ce problème représente l’optimisation du transport du poisson frais entre Nouadhibou et Nouakchott.

L’objectif est de maximiser le bénéfice total tout en respectant les contraintes de capacité.

## Fonction objectif

Maximiser :


Z = 40X + 50Y

Sous les contraintes :


\begin{array}{l}
X + Y \leq 100 \\
2X + 3Y \leq 240 \\
X + 2Y \leq 160 \\
X, Y \geq 0
\end{array}


## Méthode utilisée

Le problème est résolu avec PuLP comme problème de programmation linéaire.

---

```python

model = LpProblem("Transport_Poissons", LpMaximize)

X = LpVariable("X", lowBound=0)
Y = LpVariable("Y", lowBound=0)

model += 40 * X + 50 * Y

model += X + Y <= 100
model += 2 * X + 3 * Y <= 240
model += X + 2 * Y <= 160

model.solve()

print("TRANSPORT POISSONS")

print("Statut :", LpStatus[model.status])

print("X =", value(X))
print("Y =", value(Y))

print("Zmax =", value(model.objective))
```
